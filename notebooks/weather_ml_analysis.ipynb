```python
import pandas as pd

df = pd.read_csv("../data/processed/lagos_weather_transformed.csv")
df.head()
```

```python
df.shape
```

```python
df["date"] = pd.to_datetime(df["date"])
```

```python
from datetime import datetime

monthly_df = df.copy()

monthly_df["month"] = monthly_df["date"].dt.to_period("M")

monthly_df.head()
```

```python
monthly_weather = (
    monthly_df
    .groupby("month")
    .agg({
        "temperature_2m_max": "mean",
        "temperature_2m_mean": "mean",
        "temperature_2m_min": "mean",

        "dew_point_2m_max": "mean",
        "dew_point_2m_mean": "mean",
        "dew_point_2m_min": "mean",

        "relative_humidity_2m_max": "mean",
        "relative_humidity_2m_mean": "mean",
        "relative_humidity_2m_min": "mean",

        "surface_pressure_max": "mean",
        "surface_pressure_mean": "mean",
        "surface_pressure_min": "mean",

        "wind_speed_10m_max": "mean",
        "wind_speed_10m_mean": "mean",
        "wind_speed_10m_min": "mean",

        "precipitation_sum": "sum"
    })
    .reset_index()
)
print(monthly_weather.shape)
monthly_weather.head()
```

```python
monthly_weather["next_month_precipitation"] = (
    monthly_weather["precipitation_sum"].shift(-1)
)

monthly_weather = monthly_weather.dropna(
    subset=["next_month_precipitation"]
).reset_index(drop=True)
```

```python
print(monthly_weather.shape)
monthly_weather.head()
```

```python
target = "next_month_precipitation"

features = [
    "temperature_2m_max",
    "temperature_2m_mean",
    "temperature_2m_min",

    "dew_point_2m_max",
    "dew_point_2m_mean",
    "dew_point_2m_min",

    "relative_humidity_2m_max",
    "relative_humidity_2m_mean",
    "relative_humidity_2m_min",

    "surface_pressure_max",
    "surface_pressure_mean",
    "surface_pressure_min",

    "wind_speed_10m_max",
    "wind_speed_10m_mean",
    "wind_speed_10m_min",

    "precipitation_sum"
]

X = monthly_weather[features]
y = monthly_weather[target]
```

```python
print("X shape:", X.shape)
print("y shape:", y.shape)
```

```python
print("Missing values in X:")
print(X.isnull().sum())

print("\nMissing values in y:")
print(y.isnull().sum())
```

```python
y.isna().sum()
```

```python
X_train = X.iloc[:-24]
X_test = X.iloc[-24:]

y_train = y.iloc[:-24]
y_test = y.iloc[-24:]

print("X_train:", X_train.shape)
print("X_test:", X_test.shape)
print("y_train:", y_train.shape)
print("y_test:", y_test.shape)
```

```python
from sklearn.ensemble import RandomForestRegressor

model = RandomForestRegressor(
    n_estimators=200,
    random_state=42
)

model.fit(X_train, y_train)

y_pred = model.predict(X_test)
```

```python
print(y_pred[:10])
```

```python
y_test.head(10).values
```

```python
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
import numpy as np

mae = mean_absolute_error(y_test, y_pred)
rmse = np.sqrt(mean_squared_error(y_test, y_pred))
r2 = r2_score(y_test, y_pred)

print(f"MAE:  {mae:.2f}")
print(f"RMSE: {rmse:.2f}")
print(f"R²:   {r2:.2f}")
```

```python
feature_importance = pd.DataFrame({
    "feature": X_train.columns,
    "importance": model.feature_importances_
}).sort_values(
    by="importance",
    ascending=False
)

feature_importance
```

```python
comparison = monthly_weather.iloc[-24:][
    ["month", "precipitation_sum", "next_month_precipitation"]
].copy()

comparison["predicted_precipitation"] = y_pred

comparison["error"] = (
    comparison["predicted_precipitation"]
    - comparison["next_month_precipitation"]
)

comparison
```
